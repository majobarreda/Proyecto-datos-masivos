```python
# ==========================================
# PROYECTO: CLIMA Y CALIDAD DEL AIRE
# PRIMERA ETAPA: CARGA Y EXPLORACIÓN INICIAL
# ==========================================

# 1. Importar librerías
import pandas as pd
import requests
import matplotlib.pyplot as plt

# --- Extracción de datos vía API (Requisito Entregable 1) ---
url_api = "https://api.open-meteo.com/v1/forecast?latitude=19.4326&longitude=-99.1332&current_weather=true"
respuesta_api = requests.get(url_api)

if respuesta_api.status_code == 200:
    print("Extracción vía API exitosa:")
    print(respuesta_api.json().get('current_weather', {}))
else:
    print("Error al conectar con la API")
print("-" * 50)

# 2. Cargar dataset subiendo un nivel hacia la carpeta data
df_kaggle = pd.read_csv("../data/LA_daily_climate.csv")
df_aire = pd.read_csv("../data/LA_daily_air_quality.csv")

# 3. Mostrar las primeras filas
print("PRIMERAS FILAS DEL DATASET:")
display(df_kaggle.head())

# 4. Mostrar dimensiones
print("\nDIMENSIONES DEL DATASET:")
print("Filas:", df_kaggle.shape[0])
print("Columnas:", df_kaggle.shape[1])

# 5. Mostrar nombres de columnas
print("\nCOLUMNAS:")
print(df_kaggle.columns.tolist())

# 6. Mostrar tipos de datos
print("\nTIPOS DE DATOS:")
print(df_kaggle.dtypes)

# 7. Información general del dataset
print("\nINFORMACIÓN GENERAL:")
df_kaggle.info()

# 8. Valores faltantes
print("\nVALORES FALTANTES POR COLUMNA:")
print(df_kaggle.isnull().sum())

# 9. Registros duplicados
print("\nNÚMERO DE DUPLICADOS:")
print(df_kaggle.duplicated().sum())

# 10. Estadísticas descriptivas
print("\nESTADÍSTICAS DESCRIPTIVAS:")
display(df_kaggle.describe(include="all"))

# 11. Mostrar algunos valores únicos por columna
print("\nEJEMPLOS DE VALORES ÚNICOS:")

for columna in df_kaggle.columns:
    print(f"\n{columna}:")
    print(df_kaggle[columna].dropna().unique()[:10])
```

```python
# ==========================================
# 1. EXTRACCIÓN HISTÓRICA VÍA API (OPEN-METEO)
# ==========================================
import requests
import pandas as pd
import matplotlib.pyplot as plt

ciudades = df_kaggle[["city", "country", "latitude", "longitude"]].drop_duplicates().reset_index(drop=True)
URL_API = "https://air-quality-api.open-meteo.com/v1/air-quality"

fecha_inicio = "2023-01-01"
fecha_fin = "2024-04-20"
columnas_contaminantes = ["pm10", "pm2_5", "carbon_monoxide", "nitrogen_dioxide", "sulphur_dioxide", "ozone"]
resultados = []

print("Número de ciudades a consultar:", len(ciudades))
for _, fila in ciudades.iterrows():
    ciudad, pais = fila["city"], fila["country"]
    latitud, longitud = fila["latitude"], fila["longitude"]
    print(f"Consultando API: {ciudad}, {pais}")

    parametros = {
        "latitude": latitud,
        "longitude": longitud,
        "hourly": "pm10,pm2_5,carbon_monoxide,nitrogen_dioxide,sulphur_dioxide,ozone",
        "start_date": fecha_inicio,
        "end_date": fecha_fin,
        "timezone": "GMT",
        "domains": "cams_global"
    }

    respuesta = requests.get(URL_API, params=parametros, timeout=60)
    respuesta.raise_for_status()
    datos = respuesta.json()

    df_ciudad = pd.DataFrame(datos["hourly"])
    df_ciudad["city"], df_ciudad["country"] = ciudad, pais
    df_ciudad["latitude"], df_ciudad["longitude"] = latitud, longitud

    for columna in columnas_contaminantes:
        df_ciudad[columna] = pd.to_numeric(df_ciudad[columna], errors="coerce")

    df_ciudad["time"] = pd.to_datetime(df_ciudad["time"], utc=True)
    resultados.append(df_ciudad)

print("\nExtracción completada.")

# Consolidación a promedio diario
df_aire_horario = pd.concat(resultados, ignore_index=True)
df_aire_horario["date"] = df_aire_horario["time"].dt.floor("D")

df_aire = df_aire_horario.groupby(
    ["date", "city", "country", "latitude", "longitude"], as_index=False
)[columnas_contaminantes].mean()

# ==========================================
# 2. DIAGNÓSTICO DE DATOS DE LA API
# ==========================================
print("\n🔍 DIAGNÓSTICO DE CALIDAD DEL AIRE:")
print("Dimensiones:", df_aire.shape)
print("Valores faltantes:\n", df_aire.isnull().sum())
print("Duplicados:", df_aire.duplicated(subset=["date", "city"]).sum())

# ==========================================
# 3. UNIÓN DE DATASETS (CLIMA + CALIDAD DEL AIRE)
# ==========================================
df_kaggle["date"] = pd.to_datetime(df_kaggle["date"], utc=True)
df_aire["date"] = pd.to_datetime(df_aire["date"], utc=True)

df_completo = pd.merge(
    df_kaggle,
    df_aire,
    on=["date", "city", "country", "latitude", "longitude"],
    how="inner"
)

print("\nDATASET INTEGRADO:")
print(f"Filas: {df_completo.shape[0]} | Columnas: {df_completo.shape[1]}")

# ==========================================
# 4. PRIMERA EVIDENCIA Y GRÁFICA
# ==========================================
promedios_ciudad = df_completo.groupby("city")[
    ["temperature_2m_mean", "precipitation_sum", "wind_speed_10m_max", "pm10", "pm2_5"]
].mean().sort_values("pm2_5", ascending=False)

plt.figure(figsize=(12, 5))
promedios_ciudad["pm2_5"].plot(kind="bar", color="darkred")
plt.title("Promedio de PM2.5 por Ciudad (API + Kaggle)")
plt.xlabel("Ciudad")
plt.ylabel("Concentración PM2.5")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
ciudades = (
    df_kaggle[
        ["city", "country", "latitude", "longitude"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

URL_API = "https://air-quality-api.open-meteo.com/v1/air-quality"

fecha_inicio = "2023-01-01"
fecha_fin = "2024-04-20"

columnas_contaminantes = [
    "pm10",
    "pm2_5",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone"
]

resultados = []

print("Número de ciudades a consultar:", len(ciudades))
display(ciudades)
```

```python
for _, fila in ciudades.iterrows():

    ciudad = fila["city"]
    pais = fila["country"]
    latitud = fila["latitude"]
    longitud = fila["longitude"]

    print(f"Consultando: {ciudad}, {pais}")

    parametros = {
        "latitude": latitud,
        "longitude": longitud,
        "hourly": (
            "pm10,"
            "pm2_5,"
            "carbon_monoxide,"
            "nitrogen_dioxide,"
            "sulphur_dioxide,"
            "ozone"
        ),
        "start_date": fecha_inicio,
        "end_date": fecha_fin,
        "timezone": "GMT",
        "domains": "cams_global"
    }

    respuesta = requests.get(
        URL_API,
        params=parametros,
        timeout=60
    )

    respuesta.raise_for_status()

    datos = respuesta.json()

    df_ciudad = pd.DataFrame(datos["hourly"])

    df_ciudad["city"] = ciudad
    df_ciudad["country"] = pais
    df_ciudad["latitude"] = latitud
    df_ciudad["longitude"] = longitud

    for columna in columnas_contaminantes:
        df_ciudad[columna] = pd.to_numeric(
            df_ciudad[columna],
            errors="coerce"
        )

    df_ciudad["time"] = pd.to_datetime(
        df_ciudad["time"],
        utc=True
    )

    resultados.append(df_ciudad)

print("\nExtracción completada.")
```

```python
df_aire_horario = pd.concat(
    resultados,
    ignore_index=True
)

df_aire_horario["date"] = (
    df_aire_horario["time"].dt.floor("D")
)

df_aire = (
    df_aire_horario
    .groupby(
        [
            "date",
            "city",
            "country",
            "latitude",
            "longitude"
        ],
        as_index=False
    )[columnas_contaminantes]
    .mean()
)

print("Primeras filas de calidad del aire:")
display(df_aire.head())

print("\nDimensiones:")
print(df_aire.shape)
```
