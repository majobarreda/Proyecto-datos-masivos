```python
# ==========================================
# PROYECTO: CLIMA Y CALIDAD DEL AIRE EN LATAM
# TERCERA ETAPA: ANÁLISIS EXPLORATORIO Y VISUALIZACIÓN
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar el dataset consolidado masivo
df = pd.read_csv("../data/LA_clima_y_aire_procesado.csv")
df['date'] = pd.to_datetime(df['date'])

print(f"Dataset cargado correctamente con {df.shape[0]:,} registros y {df.shape[1]} columnas.")

# 2. Gráfica 1: Temperatura Promedio por Ciudad
plt.figure(figsize=(12, 6))
df.groupby('city')['temperature_2m_mean'].mean().sort_values().plot(kind='barh', color='teal')
plt.title('Temperatura Promedio por Ciudad en LATAM (°C)', fontsize=14, fontweight='bold')
plt.xlabel('Temperatura Media (°C)')
plt.ylabel('Ciudad')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# 3. Gráfica 2: Matriz de Correlación (Clima vs. Calidad del Aire)
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr = numeric_df.corr()

plt.figure(figsize=(10, 8))
plt.imshow(corr, cmap='coolwarm', aspect='auto')
plt.colorbar(label='Coeficiente de Correlación')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Matriz de Correlación entre Variables de Clima y Calidad del Aire', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
