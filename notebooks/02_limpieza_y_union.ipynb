```python
# ==========================================
# PROYECTO: CLIMA Y CALIDAD DEL AIRE EN LATAM
# SEGUNDA ETAPA: LIMPIEZA Y UNIÓN MASIVA
# ==========================================

import pandas as pd

# 1. Cargar datasets
df_clima = pd.read_csv("../data/LA_daily_climate.csv")
df_aire = pd.read_csv("../data/LA_daily_air_quality.csv")

# 2. Limpiar espacios invisibles en nombres (en ambos DataFrames)
for df in [df_clima, df_aire]:
    for col in ['city', 'country']:
        if col in df.columns:
            df[col] = df[col].astype(str).str.strip()

# 3. Convertir fechas a datetime estandarizando en UTC
df_clima['date'] = pd.to_datetime(df_clima['date'], utc=True)
df_aire['date'] = pd.to_datetime(df_aire['date'], utc=True)

# 4. Unir datasets masivos de forma correcta (por fecha Y ubicación)
llaves_union = ['date', 'city', 'country', 'latitude', 'longitude']
df_completo = pd.merge(df_clima, df_aire, on=llaves_union, how='inner').drop_duplicates()

# 5. Mostrar resultados
print("DATASET CONSOLIDADO MASIVO:")
print(f" - Total de filas: {df_completo.shape[0]:,}")
print(f" - Total de columnas: {df_completo.shape[1]}")
display(df_completo.head())

# 6. Guardar archivo procesado
df_completo.to_csv("../data/LA_clima_y_aire_procesado.csv", index=False)
print("\nArchivo 'data/LA_clima_y_aire_procesado.csv' guardado con éxito.")
```
