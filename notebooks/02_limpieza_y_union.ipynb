```python
# ==========================================
# PROYECTO: CLIMA Y CALIDAD DEL AIRE EN LATAM
# SEGUNDA ETAPA: LIMPIEZA Y UNIÓN MASIVA
# ==========================================

import pandas as pd

# 1. Cargar datasets
df_clima = pd.read_csv("../data/LA_daily_climate.csv")
df_aire = pd.read_csv("../data/LA_daily_air_quality.csv")

# 2. Limpiar espacios invisibles en nombres
if 'city' in df_clima.columns:
    df_clima['city'] = df_clima['city'].astype(str).str.strip()
if 'country' in df_clima.columns:
    df_clima['country'] = df_clima['country'].astype(str).str.strip()

# 3. Convertir fechas a datetime
df_clima['date'] = pd.to_datetime(df_clima['date'])
df_aire['date'] = pd.to_datetime(df_aire['date'])

# 4. Unir datasets masivos
df_completo = pd.merge(df_clima, df_aire, on='date', how='inner', suffixes=('_clima', '_aire')).drop_duplicates()

# 5. Mostrar resultados
print("✅ DATASET CONSOLIDADO MASIVO:")
print(f"   - Total de filas: {df_completo.shape[0]:,}")
print(f"   - Total de columnas: {df_completo.shape[1]}")
display(df_completo.head())

# 6. Guardar archivo procesado
df_completo.to_csv("../data/LA_clima_y_aire_procesado.csv", index=False)
print("\n💾 Archivo 'data/LA_clima_y_aire_procesado.csv' guardado con éxito.")
```
